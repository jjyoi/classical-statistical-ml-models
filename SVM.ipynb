# SVM Baseline Model

### imports

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
```

### example data

```python
X = [
    "The food was delicious",
    "Amazing service and great food",
    "I loved this restaurant",
    "The meal was excellent",
    "The food was terrible",
    "Slow service and bad food",
    "I disliked this restaurant",
    "The meal was disappointing",
]

y = [
    "positive", "positive", "positive", "positive",
    "negative", "negative", "negative", "negative",
]
```

### 1. create a held-out validation set

```python
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y,
)
```

### 2. build a preprocessing + model pipeline

```python
model = make_pipeline(
    TfidfVectorizer(),
    LinearSVC(C=1.0),
)
# model = make_pipeline(
#     StandardScaler(),
#     SVC(kernel="rbf", C=1.0, gamma="scale")
# )
```

### 3. evaluation

```python
# train model
model.fit(X_train, y_train)

# predict
y_pred = model.predict(X_val)

# evaluate 
print("Accuracy:", accuracy_score(y_val, y_pred))
print("Macro F1:", f1_score(y_val, y_pred, average="macro"))
```

### 4. new data 

```python
new_reviews = ["The food was great", "The service was terrible"]
print("Predictions:", model.predict(new_reviews))
```
